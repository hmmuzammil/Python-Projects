### 3. How much Employees are there in each Department?

```python
dept = df['Department'].value_counts()
dept
```

```python
sns.countplot(x= 'Department', data = df, palette = ['navy','yellow', 'lightgreen' ])
plt.xticks(rotation = 'vertical')
plt.show()
```

### 4. What is the average Salary of each Department?


```python
salary = df.groupby('Department')['Salary_INR'].mean()/1000
salary
```

```python
salary.plot( x = salary.index, y = salary.values,  kind = 'bar', color = 'blue')
plt.title('Average Salary  of each Department')
plt.xlabel('Department')
plt.ylabel('Average Salary in Thousands')
plt.grid()
plt.show();
```

### 5. What is the Average Salary of each Department based on the Job title?

```python
avg = df.groupby(['Department','Job_Title'])['Salary_INR'].mean()/1000
avg
```

```python
import random
num_bars = len(avg)
random_colors = [f'#{random.randint(0, 0xFFFFFF):06x}' for _ in range(num_bars)]


avg.plot(kind = 'barh', figsize = (9,5), color = random_colors )
plt.show();
```

### 13. Identify the Departments with highest Attrition rate

```python
dept_reg = df.groupby('Department')['Status'].agg(Total_Emp = 'count', Resigned = lambda x: (x =='Resigned').sum())
dept_reg
```

```python
dept_reg['Attrition_Rate'] = dept_reg['Re']/dept_reg['Total_Emp']
```
